```python
from PIL import Image, ImageOps
from pathlib import Path

photos = sorted(Path('/content/tasse/input').glob('*.jpeg'))[::3]
output = Path('/content/tasse_light/input')
output.mkdir(parents=True, exist_ok=True)

for photo in photos:
    with Image.open(photo) as img:
        img = ImageOps.exif_transpose(img)
        img.thumbnail((1600, 1600))
        img.convert('RGB').save(output / photo.name)

print(len(photos), "photos préparées")
```
