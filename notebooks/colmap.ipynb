```python
from plyfile import PlyData
import plotly.graph_objects as go
import numpy as np

# Charger le nuage de points COLMAP
path = "/content/drive/MyDrive/data/images_tasse/gaussian_model_3000/input.ply"
points = PlyData.read(path)["vertex"].data

xyz = np.column_stack([
    points["x"],
    points["y"],
    points["z"]
])

rgb = np.column_stack([
    points["red"],
    points["green"],
    points["blue"]
])

# Retirer les points les plus éloignés POUR LA VISUALISATION
# 0.70 = afficher les 70 % de points les plus proches du centre
fraction = 0.70

centre = np.median(xyz, axis=0)
distances = np.linalg.norm(xyz - centre, axis=1)

masque = distances <= np.quantile(distances, fraction)

xyz = xyz[masque]
rgb = rgb[masque]

couleurs = [
    f"rgb({r},{g},{b})"
    for r, g, b in rgb
]

# Visualisation interactive
fig = go.Figure(
    data=[
        go.Scatter3d(
            x=xyz[:, 0],
            y=xyz[:, 1],
            z=xyz[:, 2],
            mode="markers",
            marker=dict(
                size=2,
                color=couleurs,
                opacity=1
            )
        )
    ]
)

fig.update_layout(
    title="Reconstruction initiale COLMAP",
    height=750,
    paper_bgcolor="#202020",
    font=dict(color="white"),
    scene=dict(
        bgcolor="#202020",
        xaxis=dict(visible=False),
        yaxis=dict(visible=False),
        zaxis=dict(visible=False),
        aspectmode="data",
        camera=dict(
            eye=dict(x=0.9, y=0.9, z=0.6)
        )
    ),
    margin=dict(l=0, r=0, b=0, t=50)
)

fig.show()
```
